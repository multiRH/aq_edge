```python
import os
import shutil

# Define paths
drive_ssh_dir = '/content/drive/My Drive/.ssh'
local_ssh_dir = os.path.expanduser('~/.ssh')

# Create local directory
os.makedirs(local_ssh_dir, exist_ok=True)

# Copy keys from Google Drive
shutil.copy(os.path.join(drive_ssh_dir, 'id_ed25519'), os.path.join(local_ssh_dir, 'id_ed25519'))
shutil.copy(os.path.join(drive_ssh_dir, 'id_ed25519.pub'), os.path.join(local_ssh_dir, 'id_ed25519.pub'))

# Adjust permissions
!chmod 700 ~/.ssh
!chmod 600 ~/.ssh/id_ed25519
!chmod 644 ~/.ssh/id_ed25519.pub

# Verify GitHub host
!ssh-keyscan -t ed25519 github.com >> ~/.ssh/known_hosts

print("✔ SSH key successfully loaded from Google Drive and configured.")
```
