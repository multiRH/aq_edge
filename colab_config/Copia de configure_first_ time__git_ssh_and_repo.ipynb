### Step 2: Generate SSH Key and save it to Google Drive
If a key already exists in your Google Drive, this step will be skipped to prevent overwriting existing keys.

```python
import os
import shutil

# Define paths in Google Drive and locally
drive_ssh_dir = '/content/drive/My Drive/.ssh'
os.makedirs(drive_ssh_dir, exist_ok=True)

local_ssh_dir = os.path.expanduser('~/.ssh')
os.makedirs(local_ssh_dir, exist_ok=True)

private_key_drive = os.path.join(drive_ssh_dir, 'id_ed25519')
public_key_drive = os.path.join(drive_ssh_dir, 'id_ed25519.pub')

# Generate key if it does not exist yet
if not os.path.exists(private_key_drive):
    print("Generating new SSH key...")
    !ssh-keygen -t ed25519 -C "fernandoarevalon@gmail.com" -N "" -f "{private_key_drive}"
else:
    print("SSH key already exists in Google Drive.")

# Copy keys to local runtime environment
shutil.copy(private_key_drive, os.path.join(local_ssh_dir, 'id_ed25519'))
shutil.copy(public_key_drive, os.path.join(local_ssh_dir, 'id_ed25519.pub'))

# Set correct permissions
!chmod 700 ~/.ssh
!chmod 600 ~/.ssh/id_ed25519
!chmod 644 ~/.ssh/id_ed25519.pub

print("✔ Key configured successfully.")
```

### Step 3: Display Public Key (for GitHub)
Copy the output of the following cell and paste it into GitHub under **Settings -> SSH and GPG keys -> New SSH key**.

```python
with open(os.path.expanduser('~/.ssh/id_ed25519.pub'), 'r') as f:
    print("Your public SSH Key for GitHub:\n")
    print(f.read().strip())
```
